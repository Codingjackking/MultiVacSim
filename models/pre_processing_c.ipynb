```python
import pandas as pd
import os

# ✅ Set your raw data folder path
folder_path = "C:\\Users\\Lenovo\\MultiVacSim\\data\\covid"

# ✅ Where to save cleaned files
output_folder = os.path.join(folder_path, "processed")
os.makedirs(output_folder, exist_ok=True)

# 📂 Original COVID data files
selected_files = {
    "covid_hospitalization": "Weekly_Rates_of_Laboratory-Confirmed_COVID-19_Hospitalizations_from_the_COVID-NET_Surveillance_System_20250220.csv",
    "covid_vaccination": "Weekly_Cumulative_Estimated_Number_of_COVID-19_Vaccinations_Administered_in_Pharmacies_and_Physician_Medical_Offices__Adults_18_years_and_older__by_Season_and_Age_Group__United_States_20250220.csv",
    "covid_mortality": "Total_COVID-19_Deaths_since_January_1__2020_by_Age_Group__Race_Ethnicity__and_Sex_20250220.csv",
    "covid_trend": "Weekly_Parental_Intent_for_Vaccination_and_Cumulative_Percentage_of_Children_6_Months_-17_Years_Who_are_Up_to_date_with_the_COVID-19_Vaccines_by_Season__United_States_20250220.csv"
}

# 🔍 Keywords to extract relevant COVID columns
relevant_keywords = {
    "covid_hospitalization": ["covid", "hospital", "rate", "age", "state", "season"],
    "covid_vaccination": ["covid", "vaccine", "vaccination", "dose", "age", "percent"],
    "covid_mortality": ["covid", "death", "mortality", "week", "age", "state"],
    "covid_trend": ["covid", "percent", "clinic", "outpatient", "ili", "visits", "trend", "week", "positivity", "testing"]
}

# 🧠 Process each file and save the result
for category, filename in selected_files.items():
    file_path = os.path.join(folder_path, filename)
    
    if not os.path.exists(file_path):
        print(f"❌ File not found: {filename}")
        continue

    try:
        df = pd.read_csv(file_path, low_memory=False)

        # ✅ Keep columns with relevant keywords (but NOT flu or rsv)
        filtered_columns = [
            col for col in df.columns
            if any(kw in col.lower() for kw in relevant_keywords[category])
            and "flu" not in col.lower()
            and "rsv" not in col.lower()
        ]

        # Final DataFrame
        filtered_df = df[filtered_columns].dropna(how="all")

        # 📁 Save file
        output_filename = f"processed_{category}.csv"
        output_path = os.path.join(output_folder, output_filename)
        filtered_df.to_csv(output_path, index=False)

        print(f"✅ Saved: {output_filename} ({len(filtered_df.columns)} columns)")

    except Exception as e:
        print(f"❌ Error processing {filename}: {e}")
```
