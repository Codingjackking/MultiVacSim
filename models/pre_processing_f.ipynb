```python
import pandas as pd
import os

# 📁 Path to your data
folder_path = "C:\\Users\\Lenovo\\MultiVacSim\\data\\flu"

# 📄 File lists by category
flu_files = {
    "flu_vaccination": [
        "Characteristics_age.csv",
        "Medical_Conditions_age.csv",
        "Weekly_Data_Counts_by_Age.csv",
        "Weekly_Data_Percent_by_Age.csv"
    ],
    "flu_mortality": [
        "Pneumonia_Influenza_Mortality_Surveillance_from_the_National_Center_Mortality_Surveillance_System_National_Custom_Data.csv",
        "Pneumonia_Influenza_Mortality_Surveillance_from_the_National_Center_Mortality_Surveillance_System_Regional_Custom_Data.csv",
        "Pneumonia_Influenza_Mortality_Surveillance_from_the_National_Center_Mortality_Surveillance_System_State_Custom_Data.csv",
        "INFLUENZA_ASSOCIATED_PEDIATRIC_MORTALITY_Weekly.csv"
    ],
    "flu_trend": [
        "ILINet_hhs_regions.csv",
        "ILINet_national.csv",
        "ILINet_state.csv",
        "StateDatabySeason48_64.csv",
        "WHO_NREVSS_Clinical_Labs_hhs_regions.csv",
        "WHO_NREVSS_Clinical_Labs_national.csv",
        "WHO_NREVSS_Clinical_Labs_state.csv",
        "WHO_NREVSS_Combined_prior_to_2015_16_hhs_regions.csv",
        "WHO_NREVSS_Combined_prior_to_2015_16_national.csv",
        "WHO_NREVSS_Combined_prior_to_2015_16_state.csv",
        "WHO_NREVSS_Public_Health_Labs_hhs_regions.csv",
        "WHO_NREVSS_Public_Health_Labs_national.csv",
        "WHO_NREVSS_Public_Health_Labs_state.csv"
    ]
}

# 🔍 Keywords per category
relevant_keywords = {
    "flu_hospitalization": ["hospital", "rate", "age", "state", "season"],
    "flu_vaccination": ["vaccine", "vaccination", "dose", "age", "percent"],
    "flu_mortality": ["death", "mortality", "week", "age", "state"],
    "flu_trend": ["percent", "clinic", "outpatient", "ili", "visits", "trend", "week", "positivity", "testing"]
}

# 🧠 Find best file by keyword match
best_files = {}

for category, files in flu_files.items():
    keyword_list = relevant_keywords[category]
    scores = {}
    matches_per_file = {}

    for file in files:
        file_path = os.path.join(folder_path, file)
        try:
            df = pd.read_csv(file_path, nrows=5)
            columns = df.columns.str.lower()
            matched_keywords = [kw for kw in keyword_list if any(kw in col for col in columns)]
            scores[file] = len(matched_keywords)
            matches_per_file[file] = matched_keywords
        except Exception as e:
            scores[file] = -1
            matches_per_file[file] = [f"❌ Error: {str(e)}"]

    best_file = max(scores, key=scores.get)
    best_files[category] = {
        "best_file": best_file,
        "score": scores[best_file],
        "matched_keywords": matches_per_file[best_file],
        "other_files": {f: {"score": s, "matched_keywords": matches_per_file[f]} for f, s in scores.items() if f != best_file}
    }

# 📊 Output report
print("\n📊 BEST FILES PER CATEGORY (Based on Keyword Matching)\n")
for category, data in best_files.items():
    print(f"🔹 Category: {category}")
    print(f"✅ Best File: {data['best_file']}")
    print(f"   ➤ Score (matched keywords): {data['score']}")
    print(f"   ➤ Keywords used: {', '.join(relevant_keywords[category])}")
    print(f"   ➤ Matched in best file: {', '.join(data['matched_keywords']) if data['matched_keywords'] else '(none)'}")

    print("   📁 Other Files:")
    for file, info in data["other_files"].items():
        print(f"     - {file} → Score: {info['score']}, Matched: {', '.join(info['matched_keywords']) if info['matched_keywords'] else '(none)'}")

    print("-" * 80)
```
