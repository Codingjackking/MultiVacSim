```python
import pandas as pd
import os

# 📂 Path where your data lives
folder_path = "C:\\Users\\Lenovo\\MultiVacSim\\data\\flu"

# ✅ Best file per category (manually chosen earlier)
best_files = {
    "flu_vaccination": "Weekly_Data_Percent_by_Age.csv",
    "flu_mortality": "INFLUENZA_ASSOCIATED_PEDIATRIC_MORTALITY_Weekly.csv",
    "flu_trend": "ILINet_state.csv"
}

# 🔍 Category-specific flu-related keywords
relevant_keywords = {
    "flu_hospitalization": ["hospital", "rate", "age", "state", "season"],
    "flu_vaccination": ["vaccine", "vaccination", "dose", "age", "percent"],
    "flu_mortality": ["death", "mortality", "week", "age", "state"],
    "flu_trend": ["percent", "clinic", "outpatient", "ili", "visits", "trend", "week", "positivity", "testing"]
}

# 📁 Output folder for cleaned files
output_folder = os.path.join(folder_path, "processed_best_flu_files")
os.makedirs(output_folder, exist_ok=True)

# 🧼 Process and clean each file
for category, filename in best_files.items():
    file_path = os.path.join(folder_path, filename)

    try:
        # Special handling for mortality file with a metadata row
        if category == "flu_mortality":
            df = pd.read_csv(file_path, skiprows=1)
        else:
            df = pd.read_csv(file_path)

        # Keep only columns with relevant keywords
        keywords = relevant_keywords[category]
        relevant_columns = [
            col for col in df.columns
            if any(kw in col.lower() for kw in keywords)
        ]
        filtered_df = df[relevant_columns]

        # Save cleaned file
        output_filename = f"cleaned_{category}.csv"
        output_path = os.path.join(output_folder, output_filename)
        filtered_df.to_csv(output_path, index=False)

        print(f"✅ Saved: {output_filename} ({len(filtered_df.columns)} columns)")

    except Exception as e:
        print(f"❌ Error processing {filename}: {e}")
```
