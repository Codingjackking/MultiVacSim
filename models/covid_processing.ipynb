## Clean Raw COVID-19 Data Files

This cell loads all .csv files in the COVID data folder, checks if the file requires skipping header rows, and rewrites the cleaned data back in UTF-8 format. If UTF-8 fails, it tries ISO-8859-1.

```python
import pandas as pd 
import glob
import os

# Define the folder path for COVID-19 data
folder_path = "C:\\Users\\naing\\MultiVacSim\\data\\covid\\"

# Get all CSV file paths
csv_files = glob.glob(folder_path + "*.csv")

# List of files that need skipping the first 2 rows
files_to_skip_rows = [
    "Pan-Respiratory_Virus_Emergency_Department_Network_Data_2023-24_National.csv",
    "united_states_covid19_deaths_ed_visits_and_positivity_by_state.csv"
]

for file in csv_files:
    try:
        file_name = os.path.basename(file)  # Extract only the file name
        skip_rows = 2 if file_name in files_to_skip_rows else 0

        # Read the CSV file while skipping the first 2 rows for selected files
        df = pd.read_csv(
            file, 
            encoding="utf-8", 
            engine="python", 
            sep=",",  # Ensures proper comma separation
            skiprows=skip_rows,  
            quoting=3,  # Treat quotes as regular characters
            on_bad_lines="skip"  
        )
        print(f"✅ Successfully processed: {file}")

        # Replace the old file with the cleaned version
        df.to_csv(file, index=False, encoding="utf-8")

    except UnicodeDecodeError:
        try:
            df = pd.read_csv(
                file, 
                encoding="ISO-8859-1",  
                engine="python", 
                sep=",",  
                skiprows=skip_rows,  
                quoting=3,  
                on_bad_lines="skip"
            )
            print(f"✅ Successfully processed (converted encoding): {file}")

            # Replace the old file with the cleaned version
            df.to_csv(file, index=False, encoding="utf-8")

        except Exception as e:
            print(f"❌ Error processing: {file}")
            print(f"🔴 Error Message: {e}")

    except pd.errors.ParserError:
        print(f"❌ Still failing: {file}")
        print("🔴 This file might need manual inspection for formatting issues.")
```

```python
import pandas as pd
import os

# Redefine folder path after state reset
folder_path = "C:\\Users\\naing\\MultiVacSim\\data\\covid\\"

covid_hospitalization_files = [
    "Weekly_Rates_of_Laboratory-Confirmed_COVID-19_Hospitalizations_from_the_COVID-NET_Surveillance_System_20250220.csv",
    # "Weekly_Hospital_Respiratory_Data__HRD__Metrics_by_Jurisdiction__National_Healthcare_Safety_Network__NHSN__20250321.csv",
    "Rates_of_Laboratory-Confirmed_RSV__COVID-19__and_Flu_Hospitalizations_from_the_RESP-NET_Surveillance_Systems_20250220.csv",
    "united_states_covid19_deaths_ed_visits_and_positivity_by_state.csv"
]
covid_hospitalization_dfs = [pd.read_csv(folder_path + f, low_memory=False) for f in covid_hospitalization_files]
covid_hospitalization_df = pd.concat(covid_hospitalization_dfs, ignore_index=True)

# 💉 Vaccination Data (Merge multiple vaccine-related files)
covid_vaccination_files = [
    "Weekly_Cumulative_Estimated_Number_of_COVID-19_Vaccinations_Administered_in_Pharmacies_and_Physician_Medical_Offices__Adults_18_years_and_older__by_Season_and_Age_Group__United_States_20250220.csv",
    "Monthly_Cumulative_Number_and_Percent_of_Persons_Who_Received_1__2024-25_COVID-19_Vaccination_Doses__by_Age_Group__and_Jurisdiction__United_States_20250220.csv",
    "Weekly_Cumulative_COVID-19_Vaccination_Coverage__by_Season_and_Race_and_Ethnicity__Medicare_Fee-For-Service_Beneficiaries_aged__65_years_20250220.csv",
    "Weekly_Cumulative_Percentage_of_Children_Ages_6_Months_-17_Years_Who_Are_Up_to_date_with_COVID-19_Vaccines_by_Selected_Demographics_and_by_Season__United_States_20250220.csv"
]
covid_vaccination_dfs = [pd.read_csv(folder_path + f, low_memory=False) for f in covid_vaccination_files]
covid_vaccination_df = pd.concat(covid_vaccination_dfs, ignore_index=True)

# ☠️ Mortality Data (Merge death-related files)
covid_mortality_files = [
    "Total_COVID-19_Deaths_since_January_1__2020_by_Age_Group__Race_Ethnicity__and_Sex_20250220.csv",
    "Monthly_COVID-19_Death_Rates_per_100_000_Population_by_Age_Group__Race_and_Ethnicity__Sex__and_Region_with_Double_Stratification_20250220.csv",
    # "Data_Period_Death_Counts_and_Mortality_Rates_by_State.csv"
]
covid_mortality_dfs = [pd.read_csv(folder_path + f, low_memory=False) for f in covid_mortality_files]
covid_mortality_df = pd.concat(covid_mortality_dfs, ignore_index=True)

# 📊 Trend Data (Merge seasonal vaccination trends)
covid_trend_files = [
    "Weekly_Parental_Intent_for_Vaccination_and_Cumulative_Percentage_of_Children_6_Months_-17_Years_Who_are_Up_to_date_with_the_COVID-19_Vaccines_by_Season__United_States_20250220.csv",
    "Weekly_Cumulative_Doses__in_Millions__of_Influenza_Vaccines_Distributed_by_Season__United_States_20250220.csv",
    "Weekly_Hospital_Respiratory_Data__HRD__Metrics_by_Jurisdiction__National_Healthcare_Safety_Network__NHSN__20250321.csv"
]
covid_trend_dfs = [pd.read_csv(folder_path + f, low_memory=False) for f in covid_trend_files]
covid_trend_df = pd.concat(covid_trend_dfs, ignore_index=True)


# Function reused from previous code
def find_best_file(file_list, category):
    relevant_keywords = {
        "covid_hospitalization": ["hospital", "rate", "age", "state", "season"],
        "covid_vaccination": ["vaccine", "vaccination", "dose", "age", "percent"],
        "covid_mortality": ["death", "mortality", "week", "age", "state"],
        "covid_trend": ["percent", "clinic", "outpatient", "ili", "visits", "trend", "week", "positivity","testing"]
    }

    scores = {}
    for f in file_list:
        try:
            df = pd.read_csv(os.path.join(folder_path, f), nrows=5)
            columns = df.columns.str.lower().tolist()
            score = sum(any(kw in col for col in columns) for kw in relevant_keywords[category])
            scores[f] = score
        except Exception:
            scores[f] = -1  # Penalize unreadable files

    best_file = max(scores, key=scores.get)
    return best_file, [f for f in file_list if f != best_file]

# Build final output
best_files = {
    "covid_hospitalization": {},
    "covid_vaccination": {},
    "covid_mortality": {},
    "covid_trend": {}
}

best_files["covid_hospitalization"]["use"], best_files["covid_hospitalization"]["comment_out"] = find_best_file(covid_hospitalization_files, "covid_hospitalization")
best_files["covid_vaccination"]["use"], best_files["covid_vaccination"]["comment_out"] = find_best_file(covid_vaccination_files, "covid_vaccination")
best_files["covid_mortality"]["use"], best_files["covid_mortality"]["comment_out"] = find_best_file(covid_mortality_files, "covid_mortality")
best_files["covid_trend"]["use"], best_files["covid_trend"]["comment_out"] = find_best_file(covid_trend_files, "covid_trend")

best_files
```
